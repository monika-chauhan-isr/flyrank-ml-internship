## Success metric

Precision@K measures the share of pages in the top K recommendations that have the declining proxy label. A higher Precision@K is better because it means the review queue contains more pages that match the target proxy. ## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule could prioritize pages using one simple condition, such as pages with declining trends or low CTR. However, page performance depends on multiple signals at the same time, including impressions, clicks, CTR, position, search volume, content characteristics, and engagement measures. These signals can interact in ways that are difficult to capture with a single if-statement. ML can learn patterns across these signals and produce a ranked review queue, while still allowing the team to compare its performance with a simple fixed-rule baseline.
